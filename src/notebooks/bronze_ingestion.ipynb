### 1-taxi_zone

```python

# تعريف المسار المباشر للـ Volume في قاعدة بيانات bronze
CSV_VOLUME_PATH = "/Volumes/assetbundles/bronze/raw_data/zones/"

@dlt.table(
    name="bronze_taxi_zone",
    comment="bronze table for taxi zone",
  
)
def bronze_taxi_zone():
    return (
        spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "csv")
        .option("header", "true")
        .option("inferSchema", "true")  # أو ترك Auto Loader يكتشف الـ Schema
        .load(CSV_VOLUME_PATH)
        .withColumn("ingestion_time", current_timestamp())
        
    )
```

### 2-tripdata table

```python


PARQUET_VOLUME_PATH = "/Volumes/assetbundles/bronze/raw_data/trips/"

@dlt.table(
    name="bronze_taxi_trip",
    comment="bronze table for taxi trip",
   
)
def bronze_taxi_trip():
    return (
        spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "parquet")
        
        .load(PARQUET_VOLUME_PATH)

        .withColumn("ingestion_time", current_timestamp())
        
    )
```
